## Initial Dataset Exploration - Observations

### Dataset Overview
- The dataset contains 23,783 mutual fund records and 298 variables.
- Most variables are numerical (277 float columns), while 21 are categorical/text columns.

### Data Quality
- The `initial_investment` column contains missing values and will require further investigation.
- Additional missing-value analysis will be performed across the full dataset.

### Initial Findings
- The `initial_investment` distribution appears to be highly right-skewed.
- The mean initial investment (~$5.18M) is much higher than the median ($2,500), suggesting that a relatively small number of funds with very large investment requirements may be influencing the average.

### Questions for Further Analysis

## Missing Values - Observations

- Several variables contain missing values, but the extent of missingness varies considerably across the dataset.
- Some variables, such as `last_cap_gain`, are missing values for nearly all observations and may have limited analytical value.
- Historical return variables from the early 2000s also contain many missing values. A possible explanation is that many mutual funds in the dataset were created after those years and therefore do not have historical return data for that period.
- Additional investigation is required before deciding whether any variables should be removed or excluded from specific analyses.
