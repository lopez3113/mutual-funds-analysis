### Business Problem

A structured data analysis can uncover relationships among key fund characteristics and generate insights that support more informed investment decisions.

## Missing Values Assessment (July 24, 2026)

### Observations

- Missing values are present across several variables, but the extent of missingness varies considerably throughout the dataset.
- Variables with the highest percentages of missing values are primarily related to historical fund returns, sales load information, and optional fund characteristics.

- Most funds are classified as **Large-cap**, followed by **Medium-cap** and **Small-cap** funds.
- Morningstar overall ratings are concentrated around **3 and 4 stars**, with fewer funds receiving the highest or lowest ratings.

### 3. Missing Value Strategy (Version 1)

| Situation | Planned Action | Reason |
|-----------|----------------|--------|
| Missing values in key analysis variables (e.g., `initial_investment`, `fund_return_1year`) | Keep the columns and decide how to handle missing values based on the specific analysis. Some rows may be excluded only when necessary. | These variables are essential to answering the project's business questions. |
| Columns with extremely high missingness (approximately >95%) | Consider removing the entire column after confirming it has little analytical value. | These columns contain too little information to support meaningful analysis. |
| Historical return variables with many missing values | Keep the columns for now. Missing values are likely due to newer funds lacking historical performance. | The missingness reflects the nature of the data rather than a data quality problem. |
| Optional fund characteristics (e.g., sales load information, ESG metrics) | Evaluate each group individually before deciding whether to keep or remove it. | These variables may be useful for future analyses even if they are not part of the initial project scope. |
| Missing categorical values | Investigate whether the missing values represent unavailable information or a meaningful category before taking action. | Avoid making assumptions about missing categorical data. |
| Missing numerical values | Do not automatically replace missing values. Choose an appropriate treatment based on the variable and the analysis being performed. | Different numerical variables require different handling strategies. |
